```python
import pandas as pd
import numpy as np
import os

# Merge
seeds = [2411, 24, 200]
dfs = []
for seed in seeds:
    df = pd.read_csv(f"output_data_[{seed}]/summary_raw.csv")
    df["seed"] = seed
    dfs.append(df)

merged = pd.concat(dfs, ignore_index=True)

# Tạo output folder
os.makedirs("output_final", exist_ok=True)
merged.to_csv("output_final/summary_raw.csv", index=False)

# Aggregate
rows = []
for scenario in merged["scenario"].unique():
    for strategy in merged["strategy"].unique():
        subset = merged[(merged["scenario"] == scenario) & (merged["strategy"] == strategy)]
        rows.append({
            "Scenario": scenario,
            "Strategy": strategy,
            "Post-Fill": f"{subset['post_fill'].mean():.3f} ± {subset['post_fill'].std():.3f}",
            "Min Fill": f"{subset['min_fill'].mean():.3f} ± {subset['min_fill'].std():.3f}",
            "Final Cost": f"{subset['final_cost'].mean():.0f} ± {subset['final_cost'].std():.0f}",
            "Bullwhip": f"{subset['bullwhip'].mean():.3f} ± {subset['bullwhip'].std():.3f}",
            "TTR": f"{subset['ttr'].mean():.1f} ± {subset['ttr'].std():.1f}",
        })

agg_df = pd.DataFrame(rows)
agg_df.to_csv("output_final/summary_aggregated.csv", index=False)

print(agg_df.to_string(index=False))
print("\n✓ Saved to output_final/")
```
